```python
import dotenv
import os
import requests
dotenv.load_dotenv()
FEC_API_KEY = os.getenv('FEC_API_KEY')
botname = 'contrans'
version = '2026'
website = 'https://github.com/jkropko/contrans2026'
email = 'jkropko@virginia.edu'
useragent = f'{botname}/{version} ({website}; {email}) python-requests/{requests.__version__}'
headers = {'User-Agent': useragent}
```

```python
root = 'https://api.open.fec.gov'
endpoint = '/v1/schedules/schedule_e'
fec_results_list = []

params = {'api_key': FEC_API_KEY,
          'candidate_id': fec_list[20],
          'per_page': 100,
          'cycle': '2026'}

newresults = 1
while newresults > 0:
    r = requests.get(root + endpoint, params = params, headers = headers)
    print(r.json())
    newresults_list = r.json()['results']
    newresults = len(newresults_list)
    fec_results_list.extend(newresults_list)
    if newresults > 0:
        params['last_expenditure_date'] = r.json()['pagination']['last_indexes']['last_expenditure_date']
        params['last_index'] = r.json()['pagination']['last_indexes']['last_index']
```
