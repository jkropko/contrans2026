```python
import numpy as np
import pandas as pd
import dotenv
import os
import psycopg
from sqlalchemy import create_engine
```

```python
dotenv.load_dotenv()
POSTGRES_PASSWORD = os.getenv('POSTGRES_PASSWORD')
```

## First, we will set up an empty contrans database

```python
dbserver = psycopg.connect(
    user = 'postgres',
    password = POSTGRES_PASSWORD,
    host = 'localhost',
    port = '5432')
dbserver.autocommit = True
    
```

```python
cursor = dbserver.cursor()
```

```python
try:
    cursor.execute('CREATE DATABASE contrans')
except:
    cursor.execute('DROP DATABASE contrans')
    cursor.execute('CREATE DATABASE contrans')
```

## Then we will add our processed data to this DB

```python
bills = pd.read_csv('../data/processed/bills.csv')
committee_members = pd.read_csv('../data/processed/committee_members.csv')
committees = pd.read_csv('../data/processed/committees.csv')
crosswalk = pd.read_csv('../data/processed/crosswalk.csv')
expenditures = pd.read_csv('../data/processed/expenditures.csv')
members = pd.read_csv('../data/processed/members.csv')
terms = pd.read_csv('../data/processed/terms.csv')
vote_compare = pd.read_csv('../data/processed/vote_compare.csv')
```

```python
#SQL Alchemy engine

dbms = 'postgresql'
package = 'psycopg'
user = 'postgres'
password = POSTGRES_PASSWORD
host = 'localhost'
port = '5432'
db = 'contrans'

engine = create_engine(f'{dbms}+{package}://{user}:{password}@{host}:{port}/{db}')
```

```python
bills.to_sql('bills', con=engine, chunksize=1000, index=False, if_exists='replace')
committee_members.to_sql('committee_members', con=engine, chunksize=1000, index=False, if_exists='replace')
committees.to_sql('committees', con=engine, chunksize=1000, index=False, if_exists='replace')
crosswalk.to_sql('crosswalk', con=engine, chunksize=1000, index=False, if_exists='replace')
expenditures.to_sql('expenditures', con=engine, chunksize=1000, index=False, if_exists='replace')
members.to_sql('members', con=engine, chunksize=1000, index=False, if_exists='replace')
terms.to_sql('terms', con=engine, chunksize=1000, index=False, if_exists='replace')
vote_compare.to_sql('vote_compare', con=engine, chunksize=1000, index=False, if_exists='replace')
```

## Then we will SQL at it

```python
myquery = '''
SELECT *
FROM members
WHERE state='Virginia'
'''

pd.read_sql_query(myquery, con=engine)
```
