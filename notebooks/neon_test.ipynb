```python
import numpy as np
import pandas as pd
import psycopg
import dotenv
import os
from sqlalchemy import create_engine
import duckdb
import sqlite3
```

```python
dotenv.load_dotenv()
db_url = os.getenv('NEON_DATABASE_URL_UNPOOLED')
db_url = db_url.replace('postgresql://', 'postgresql+psycopg://')
engine = create_engine(db_url, pool_pre_ping=True)
```

```python
myquery = '''
SELECT *
FROM members
'''
pd.read_sql_query(myquery, con=engine)
```

# DuckDB

```python
#os.chdir("../data/processed")
```

```python
db = duckdb.connect("contrans.duckdb")
```

```python
tables = ['bills', 'committee_members', 'committees', 'crosswalk', 
          'expenditures', 'members', 'terms', 'vote_compare']
for t in tables:
    myquery = f'''
        CREATE OR REPLACE TABLE {t} AS
        SELECT * FROM read_csv('{t}.csv')
    '''
    db.execute(myquery)
```

```python
myquery = '''
SELECT *
FROM members
WHERE state='Virginia'
'''

db.execute(myquery).df()
```

```python
db.close()
```

# SQlite

```python
db = sqlite3.connect('contrans.db')
```

```python
tables = ['bills', 'committee_members', 'committees', 'crosswalk', 
          'expenditures', 'members', 'terms', 'vote_compare']
for t in tables:
    mydf = pd.read_csv(f'{t}.csv')
    mydf.to_sql(f'{t}', con=db, index=False, chunksize=1000, if_exists='replace')
```

```python
myquery = '''
SELECT *
FROM members
WHERE state='Virginia'
'''

pd.read_sql_query(myquery, con=db)
```
